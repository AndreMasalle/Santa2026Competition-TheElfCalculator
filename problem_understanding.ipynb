
# Puzzle Contoh

**Angka awal:** `[3, 5]`
**Target:** `[15, 16, 240, 241]`
**value_limit:** `1000`

Di awal, register 0 berisi 3 dan register 1 berisi 5. Sisanya kosong.

## Solusi (5 Instruksi)

| No. | Instruksi | Artinya | Hasil | Target yang Ter-cover |
|---|---|---|---|---|
| 1 | `["*", 0, 1, 2]` | reg0 × reg1 → reg2 | 3 × 5 = 15 | 15 |
| 2 | `["/", 0, 0, 3]` | reg0 ÷ reg0 → reg3 | 3 ÷ 3 = 1 | (perantara) |
| 3 | `["+", 2, 3, 4]` | reg2 + reg3 → reg4 | 15 + 1 = 16 | 15, 16 |
| 4 | `["*", 2, 4, 5]` | reg2 × reg4 → reg5 | 15 × 16 = 240 | 15, 16, 240 |
| 5 | `["+", 5, 3, 6]` | reg5 + reg3 → reg6 | 240 + 1 = 241 | Semua |

## Keadaan Register di Akhir

```text
reg0 = 3
reg1 = 5
reg2 = 15
reg3 = 1
reg4 = 16
reg5 = 240
reg6 = 241
```

## Apa yang Bisa Dipelajari dari Contoh Ini?

### 1. Lower Bound

Lower bound = 4 (ada 4 target yang belum ada di awal), tetapi kita membutuhkan 5 instruksi.

Selisihnya karena angka 1 bukan target, tetapi berguna sebagai alat. Angka 1 dibuat sekali (`3 ÷ 3`) lalu dipakai dua kali, untuk menghasilkan 16 dan 241.

Inilah yang kumaksud dengan **perantara yang dipakai ulang**.

### 2. Target Dipakai sebagai Bahan Target Lain

240 dibuat dari `15 × 16`, yaitu dua target sebelumnya.

Itu murah (1 instruksi) karena bahannya sudah ada.

### 3. Pola "Tetangga Dekat"

- 241 = 240 + 1
- 16 = 15 + 1

Di puzzle asli, target seperti 10973744 dan 10974352 (selisih 608) kemungkinan bekerja sama: kalau 608 ada di register, target kedua tinggal 1 langkah dari yang pertama.

## Contoh Langkah yang Ilegal

- `["-", 0, 1, x]` → 3 − 5 negatif, dilarang (pengurangan hanya jika angka pertama lebih besar).
- `["/", 1, 0, x]` → 5 ÷ 3 tidak habis, dilarang.
- `["*", 5, 5, x]` → 240 × 240 = 57600 melebihi limit 1000, dilarang.
- `["+", 0, 9, x]` → reg9 belum pernah ditulis, dilarang.

## Hubungan dengan Puzzle Asli

Idenya sama persis, hanya skalanya jauh lebih besar: angka awal kecil (misalnya 2, 9, 12), target sampai triliunan, dan sekitar 30 target.

Tantangannya adalah menemukan "angka 1"-nya, yaitu perantara murah yang bisa menjembatani banyak target sekaligus.
